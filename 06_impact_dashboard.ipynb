# 06 — Impact Analysis & KPI Dashboard
**AI-Driven Waste Collection & Route Optimization**

Full quantified comparison: Traditional vs AI-optimized system across
financial, environmental, and operational dimensions.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import warnings
warnings.filterwarnings('ignore')

# === KPI Data (from presentation) ===
kpis = {
    'Collection Trips Reduction (%)': {'Traditional': 100, 'AI-Optimized': 65},
    'Fuel Cost Savings (%)':          {'Traditional': 100, 'AI-Optimized': 72},
    'CO2 Emissions (relative)':       {'Traditional': 100, 'AI-Optimized': 70},
    'Overflow Events (relative)':     {'Traditional': 100, 'AI-Optimized': 15},
    'On-Time Collection Rate (%)':    {'Traditional': 72,  'AI-Optimized': 97},
    'Fleet Downtime (relative)':      {'Traditional': 100, 'AI-Optimized': 78},
    'Citizen NPS (relative)':         {'Traditional': 100, 'AI-Optimized': 140},
}

print('KPI Dashboard initialized')
```

```python
fig = plt.figure(figsize=(16, 12))
gs = gridspec.GridSpec(3, 3, figure=fig, hspace=0.45, wspace=0.35)
fig.suptitle('AI Waste Management — KPI Dashboard\nTraditional vs AI-Optimized System', 
              fontsize=15, fontweight='bold', y=0.98)

# --- (0,0) Collection trips over 5 years ---
ax = fig.add_subplot(gs[0, 0])
years  = [1, 2, 3, 4, 5]
trad_c = [1200, 1230, 1260, 1290, 1345]
ai_c   = [780,  700,  650,  610,  988-357]
x = np.arange(len(years))
ax.bar(x - 0.2, trad_c, 0.4, label='Traditional', color='#FF6B6B', alpha=0.9)
ax.bar(x + 0.2, ai_c,   0.4, label='AI-Optimized', color='#4CAF50', alpha=0.9)
ax.set_title('Annual Cost ($K) — 5 Year', fontsize=9, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels([f'Yr {y}' for y in years])
ax.legend(fontsize=7)
ax.grid(True, alpha=0.3, axis='y')
ax.set_ylabel('$K')

# --- (0,1) CO2 reduction ---
ax = fig.add_subplot(gs[0, 1])
months = list(range(1, 25))
co2_trad = [890 + np.random.normal(0, 20) for _ in months]
co2_ai   = [890 * (1 - 0.005*m) + np.random.normal(0, 15) for m in months]
ax.plot(months, co2_trad, 'r-', linewidth=1.5, label='Traditional', alpha=0.8)
ax.plot(months, co2_ai,   'g-', linewidth=1.5, label='AI-Optimized', alpha=0.8)
ax.fill_between(months, co2_ai, co2_trad, alpha=0.15, color='green', label='Savings')
ax.set_title('Monthly CO2 Emissions (kg/day)', fontsize=9, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('kg CO2/day')
ax.legend(fontsize=7)
ax.grid(True, alpha=0.3)

# --- (0,2) ROI cumulative ---
ax = fig.add_subplot(gs[0, 2])
invest = 480  # $K initial investment
ann_save = 357  # $K/yr savings
cum_save = [ann_save * (m/12) - invest for m in range(1, 37)]
ax.plot(range(1, 37), cum_save, color='#9C27B0', linewidth=2)
ax.axhline(0, color='grey', linestyle='--', linewidth=1)
breakeven = next(i+1 for i, v in enumerate(cum_save) if v >= 0)
ax.axvline(breakeven, color='green', linestyle=':', linewidth=1.5,
           label=f'Break-even M{breakeven}')
ax.fill_between(range(1, 37), cum_save, 0,
                where=[v>=0 for v in cum_save], alpha=0.2, color='green')
ax.fill_between(range(1, 37), cum_save, 0,
                where=[v< 0 for v in cum_save], alpha=0.2, color='red')
ax.set_title('Cumulative ROI ($K)', fontsize=9, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Net Return ($K)')
ax.legend(fontsize=7)
ax.grid(True, alpha=0.3)

# --- (1,:) KPI comparison bars ---
ax = fig.add_subplot(gs[1, :])
labels = list(kpis.keys())
trad_vals = [kpis[k]['Traditional'] for k in labels]
ai_vals   = [kpis[k]['AI-Optimized'] for k in labels]
x = np.arange(len(labels))
ax.bar(x - 0.2, trad_vals, 0.4, label='Traditional', color='#FF6B6B', alpha=0.9)
ax.bar(x + 0.2, ai_vals,   0.4, label='AI-Optimized', color='#4CAF50', alpha=0.9)
for i, (t, a) in enumerate(zip(trad_vals, ai_vals)):
    delta = a - t
    color = 'green' if (('Overflow' in labels[i] or 'Emissions' in labels[i] 
                         or 'Trips' in labels[i] or 'Fuel' in labels[i] 
                         or 'Downtime' in labels[i]) and delta < 0) or \
                       (('NPS' in labels[i] or 'On-Time' in labels[i]) and delta > 0) else 'red'
    ax.annotate(f'{abs(delta):.0f}', xy=(i+0.2, a+1), fontsize=7, color=color, ha='center')
ax.set_title('Key Performance Indicators — Traditional vs AI', fontsize=10, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels([l.replace(' (%)', '').replace(' (relative)', ''), 
                    for l in labels], rotation=15, ha='right', fontsize=8)
ax.legend()
ax.grid(True, alpha=0.3, axis='y')

# --- (2,0) On-time rate improvement ---
ax = fig.add_subplot(gs[2, 0])
vals = [72, 97]
bars = ax.bar(['Traditional','AI-Optimized'], vals,
               color=['#FF6B6B','#4CAF50'], alpha=0.9, width=0.5)
for bar, v in zip(bars, vals):
    ax.text(bar.get_x()+bar.get_width()/2, v+0.5, f'{v}%',
            ha='center', fontweight='bold', fontsize=11)
ax.set_title('On-Time Collection Rate', fontsize=9, fontweight='bold')
ax.set_ylim(0, 115)
ax.axhline(100, color='grey', linestyle=':', alpha=0.5)
ax.grid(True, alpha=0.3, axis='y')

# --- (2,1) Overflow events ---
ax = fig.add_subplot(gs[2, 1])
vals = [100, 15]
bars = ax.bar(['Traditional','AI-Optimized'], vals,
               color=['#FF6B6B','#4CAF50'], alpha=0.9, width=0.5)
for bar, v in zip(bars, vals):
    ax.text(bar.get_x()+bar.get_width()/2, v+1, f'{v}',
            ha='center', fontweight='bold', fontsize=11)
ax.set_title('Overflow Events (relative)', fontsize=9, fontweight='bold')
ax.set_ylabel('Relative count')
ax.grid(True, alpha=0.3, axis='y')

# --- (2,2) Summary KPI table ---
ax = fig.add_subplot(gs[2, 2])
ax.axis('off')
summary = [
    ['Metric', 'Value'],
    ['Trip reduction', '23-35%'],
    ['Fuel savings', '28%'],
    ['CO2 reduction', '30%'],
    ['Overflow eliminated', '85%'],
    ['ROI period', '14-18 mo'],
    ['5yr ROI multiplier', '3.2x'],
    ['Model accuracy', '87-93%'],
]
tbl = ax.table(cellText=summary[1:], colLabels=summary[0],
               loc='center', cellLoc='center')
tbl.auto_set_font_size(False)
tbl.set_fontsize(8)
tbl.scale(1.2, 1.4)
for (r,c), cell in tbl.get_celld().items():
    if r == 0:
        cell.set_facecolor('#1a237e')
        cell.set_text_props(color='white', fontweight='bold')
    else:
        cell.set_facecolor('#E8F5E9' if c == 1 else '#F5F5F5')
ax.set_title('Summary KPIs', fontsize=9, fontweight='bold', pad=60)

plt.savefig('../outputs/06_kpi_dashboard.png', dpi=150, bbox_inches='tight')
plt.show()
print('Dashboard saved.')
```
